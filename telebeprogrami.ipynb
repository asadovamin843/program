```python
say=0
fayl=open('telebeler.txt','w')
while True:
  ad=input()
  if ad.upper()=='DONE':
    break
  bal=int(input())
  fayl.write(ad+' '+str(bal)+'\n')
  if bal>=51:
    say=say+1
    print('kecdi')
  else:
      print('kesildi')
fayl.close()
print('kecen telebe sayi:',say)
```
